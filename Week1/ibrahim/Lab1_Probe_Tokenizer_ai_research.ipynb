```python
# pip install -U transformers accelerate
# Load model directly
from transformers import AutoProcessor, AutoModelForMultimodalLM

processor = AutoProcessor.from_pretrained("google/gemma-4-E2B-it")
model = AutoModelForMultimodalLM.from_pretrained("google/gemma-4-E2B-it", device_map="auto")
```

```python
# Probe Tokenizer for a message, the message is encoded and 
# produces a Token ids and then map the ids to it actual token to 
# display the token(text) with their respective ids
```

```python
def probing_tokenizer(text):
  tokenizer = processor.tokenizer
  encoder = tokenizer(
      text,
      add_special_tokens=False,
      return_offset_mapping=True
  )
  ids = encoder['input_ids']
  tokens = tokenizer.convert_ids_to_tokens(ids)

  print(f"Text to probe: {text}")
  print(f"Length of the probe text {len(text)}")
  print(f"Total length of token ids {len(ids)}")

  for i, (id, token) in enumerate(zip(ids, tokens)):
    print(f"{i:>3} -> Token ID: {id} -> Token: {token}")
```

```python
probing_tokenizer("Hello World")
```

```python
probing_tokenizer("Flow Research")
```

```python
probing_tokenizer("The boy should not be here")
```
